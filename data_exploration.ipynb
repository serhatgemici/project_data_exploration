```python
col_list_2001 = df_2001.columns.tolist()
col_list_2002 = df_2002.columns.tolist()
col_list_2003 = df_2003.columns.tolist()
col_list_2004 = df_2004.columns.tolist()
col_list_2005 = df_2005.columns.tolist()
col_list_2006 = df_2006.columns.tolist()
col_list_2007 = df_2007.columns.tolist()
col_list_2008 = df_2008.columns.tolist()
col_list_2009 = df_2009.columns.tolist()
col_list_2010 = df_2010.columns.tolist()
col_list_2011 = df_2011.columns.tolist()
col_list_2012 = df_2012.columns.tolist()
col_list_2013 = df_2013.columns.tolist()
col_list_2014 = df_2014.columns.tolist()
col_list_2015 = df_2015.columns.tolist()
```

```python
fig = plt.figure(figsize=(12, 6))
ax = fig.add_subplot(111)
ax.bar(range(2001, 2016), [len(col_list_2001), len(col_list_2002), len(col_list_2003), len(col_list_2004), 
                           len(col_list_2005), len(col_list_2006), len(col_list_2007), len(col_list_2008), 
                           len(col_list_2009), len(col_list_2010), len(col_list_2011), len(col_list_2012), 
                           len(col_list_2013), len(col_list_2014), len(col_list_2015)])
ax.set_xlabel("Year")
ax.set_ylabel("Number of Columns")
ax.set_title("Number of Columns per Year");
```

```python
all_cols = col_list_2001 + col_list_2002 + col_list_2003 + col_list_2004 + col_list_2005 + col_list_2006 + col_list_2007 + \
col_list_2008 + col_list_2009 + col_list_2010 + col_list_2011 + col_list_2012 + col_list_2013 + col_list_2014 + col_list_2015

freq = pd.Series(all_cols).value_counts().reset_index()
freq.columns = ["column_name", "count"]

freq.head(20)

```

```python
top_n = 20
top_freq = freq.head(top_n)

plt.figure(figsize=(10, 6))
plt.barh(top_freq["column_name"], top_freq["count"])
plt.gca().invert_yaxis()  # biggest on top
plt.xlabel("Number of years the column name is present")
plt.ylabel("Column name")
plt.title("Most common columns (2001-2015)");
```
