```python
# 1. We define the aggregation rules to be applied to each column
agg_rules = {
    'Ewltp (g/km)': 'mean',          # Continuous Numeric: Average CO2 emissions
    'Mt': 'mean',                    # Continuous Numeric: Average test weight
    'm (kg)': 'mean',                # Continuous Numeric: Average curb weight
    'ec (cm3)': 'mean',              # Continuous Numeric: Average engine displacement
    'Electric range (km)': 'mean',   # Continuous Numeric: Average electric range
    'r': 'sum',                      # Registration Count: Total number of vehicles sold from that family
    
    # Categorical Columns: Usually the same within the group, we take the first value to be safe
    'Mh': 'first',                   # Standard Manufacturer Name
    'Mp': 'first',                   # Manufacturer Pool
    'Fm': 'first',                   # Fuel Mode (E, P, H, M etc.)
    'Ct': 'first'                    # Vehicle Category (M1, M1G etc.)
}

# 2. We are performing the grouping process
df_grouped = df_de_fr_2024.groupby('RLFI').agg(agg_rules).reset_index()

# 3. We are checking the size of the new dataset
print(f"Old Row Count: {len(df_de_fr_2024)}")
print(f"New Grouped Row Count: {len(df_grouped)}")
```

What Will Change in Your Model After This Grouping? The Advantage of Weighted Linear Regression: Each row now represents not a single vehicle, but a "family of vehicles." When training your model (e.g., a LinearRegression or XGBoost), you can provide the r column (total registration count) to the model as a sample weight. This ensures that popular vehicle families with high market share are weighted more accurately in the model:
