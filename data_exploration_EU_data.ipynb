## Let's continue with **"VFN"** column;
### Vehicles approved after January 1, 2018, have a standard 24-character code structure that includes the manufacturer code (WMI) and the model year.

### **Format:** IP-TA-WMI-yyyy-nnnn

### **Components: **
### **IP:** Abbreviation for the Interpolation Family. 
### **TA:** Type Approval Authority (e.g., E1 for Germany, E2 for France).
### **WMI:** World Manufacturer Identifier (ISO standard manufacturer code). 
### **yyyy:** Year of approval. 
### **nnnn:** 4-digit family number.

## Let's check the data in **Tan** column for 2024

```python
pd.set_option("display.max_rows", None)
tan_pct = (
    df_2024["Tan"]
    .value_counts(normalize=True)
    .mul(100)
    .reset_index()
)
tan_pct.columns = ["Tan", "percentage"]
print(tan_pct.head(20))
```

```python
# TAN regex structure compliant with official EU standards
# TAN regex structure 100% compliant with EU standards (supports 4- or 5-part structures)
# Description: 
# The regex includes ALL official exceptions under EU standards (KS, NIV, IVA, 2-digit years)
# Updates made:
# (?i)                  -> Case-insensitive
# [A-Z]{0,3}\d{2,4}/    -> Allows the year portion to be 2 digits (e.g., 98/14) and to start with letters such as 'KS'
# (?:E\d{1,3}\*NIV|IVA|IV) -> Directly accepts single/special validations (E1*NIV, IVA, IV)
std_4_or_5_parts = r"e\d{1,3}\*(?:[A-Z]{1,3})?\d{2,4}/\d{2,4}\*(?:(?:[A-Z]{1,3})?\d{2,4}/\d{2,4}\*)?\d{4,6}\*\d{2,3}"
national_hybrid  = r"(?:e\d{1,3}\*)?(?:NIV|IV|NKS)(?:\*[0-9/]+|[0-9/]*)(?:[\*-]?[A-Z0-9\s-]*)?"
single_approvals = r"IVA|IV"

tan_regex = f"(?i)^\\s*(?:{std_4_or_5_parts}|{national_hybrid}|{single_approvals})\\s*$"


# Data in the correct format
tan_correct = df_2024[df_2024["Tan"].str.match(tan_regex, na=False)]

# Data in an incorrect or corrupted format
tan_corrupt = df_2024[~df_2024["Tan"].str.match(tan_regex, na=False) & df_2024["Tan"].notna()]

print(f"Correct Tan rows: {len(tan_correct)}")
print(f"Corrupt Tan rows: {len(tan_corrupt)}")

# A quick overview of the most common types of malformed formats, using dashes and asterisks:
patterns_tan = df_2024["Tan"].str.extract(r"^(.*?)\*").dropna().iloc[:, 0].value_counts()
print("\nThe distribution up to the first star (useful for spotting typos like 'E1', 'e1 ', etc.):")
print(patterns_tan.head(10))
```

```python
# Let's just look at the characters up to the first asterisk in the 98,320 corrupted lines
patterns_real_corrupt = tan_corrupt["Tan"].str.extract(r"^(.*?)\*").dropna().iloc[:, 0].value_counts()

print("Distribution up to the first star in the actual corrupted data:")
print(patterns_real_corrupt.head(15))

# If there are any inputs that do not contain any asterisks (*), the code above will not catch them.
# To see completely invalid inputs that do not contain asterisks:
no_star_corrupt = tan_corrupt[~tan_corrupt["Tan"].str.contains(r"\*", na=False)]["Tan"].value_counts()
print("\nCompletely invalid entries that contain no stars:")
print(no_star_corrupt.head(15))
```

```python
# 1. Let's analyze how many stars it contains (According to the standard, there should be 3 or 4 stars)
star_counts = tan_corrupt["Tan"].str.count(r"\*").value_counts()
print("Star Number Distribution (in Corrupted Data):")
print(star_counts)

# 2. Let’s print out some raw samples from the corrupted data containing the star and examine them visually
print("\nSample data rows that contain an asterisk but do not conform to the format:")
print(tan_corrupt[tan_corrupt["Tan"].str.contains(r"\*", na=False)]["Tan"].head(20))
```

```python
# Regex for Column T
# - (?i)^             -> The case-insensitive flag was placed at the very beginning, followed by a newline
# - [A-Z0-9(),/_\.+-] -> Allowed characters (the hyphen was placed at the end to prevent errors)
# - (?:\s+[A-Z0-9(),/_\.+-]+)* -> Allows for spaces between words (e.g., P7 MONOCAB C)
t_regex_final = r"(?i)^\s*[A-Z0-9(),/_\.+-]+(?:\s+[A-Z0-9(),/_\.+-]+)*\s*$"

# Step 1: Let's apply the regex filter (it preserves valid code containing internal whitespace)
t_correct = df_2024[df_2024["T"].str.match(t_regex_final, na=False)]
t_corrupt = df_2024[~df_2024["T"].str.match(t_regex_final, na=False) & df_2024["T"].notna()]

# Step 2: Let’s definitively filter out junk text that is longer than 25 characters and causes data corruption
true_correct_mask = df_2024["T"].str.match(t_regex_final, na=False) & (df_2024["T"].str.len() <= 25)

t_correct_clean = df_2024[true_correct_mask]
t_corrupt_clean = df_2024[~true_correct_mask & df_2024["T"].notna()]

print(f"Correct T rows: {len(t_correct_clean)}")
print(f"Corrupt/Suspicious T rows (Junk Data): {len(t_corrupt_clean)}")


# 1. If there are any, let's look at the first 20 examples of those suspicious lines that don't conform to the format:
print("\nExamples of T that appear formally questionable or flawed:")
print(t_corrupt["T"].head(20))

# Let's capture data that is too long to be a type code (e.g., an error message or a long string of text):
print("\nT entries that are longer than 15 characters and could cause data corruption:")
print(df_2024[df_2024["T"].str.len() > 15]["T"].value_counts().head(10))
```

## Let's check the data in **m (kg)** column for 2024

```python
# 1. Basic Descriptive Statistics of the clean numeric column
print("=== Descriptive Statistics for m (kg) ===")
print(df_2024["m (kg)"].describe())

# 2. Statistical Outlier Detection using the IQR (Interquartile Range) Method
Q1 = df_2024["m (kg)"].quantile(0.25)
Q3 = df_2024["m (kg)"].quantile(0.75)
IQR = Q3 - Q1

# Define standard statistical outlier boundaries
lower_bound_iqr = Q1 - 1.5 * IQR
upper_bound_iqr = Q3 + 1.5 * IQR

print("\n=== Statistical Outlier Boundaries (IQR Method) ===")
print(f"IQR Lower Bound: {lower_bound_iqr:.2f} kg")
print(f"IQR Upper Bound: {upper_bound_iqr:.2f} kg")

# 3. Domain-Specific (Physical) Outlier Detection for Passenger Cars (EEA Dataset)
# Physically, mass in running order for modern M1/N1 vehicles is almost never under 600kg or over 4000kg
physical_lower_limit = 600
physical_upper_limit = 4000

# Count how many rows violate statistical and physical limits
iqr_outliers = df_2024[(df_2024["m (kg)"] < lower_bound_iqr) | (df_2024["m (kg)"] > upper_bound_iqr)]
physical_outliers = df_2024[(df_2024["m (kg)"] < physical_lower_limit) | (df_2024["m (kg)"] > physical_upper_limit)]

print("\n=== Outlier Counts ===")
print(f"Total rows with valid numeric mass: {df_2024['m (kg)'].notna().sum()}")
print(f"Statistical Outliers (IQR): {len(iqr_outliers)} rows")
print(f"Physical Outliers (<600kg or >4000kg): {len(physical_outliers)} rows")

# 4. View samples of extreme low and extreme high values to catch data entry bugs (e.g., 0 kg or 9999 kg)
print("\n=== Sample of Lowest 10 Unique Weights (Potential Bugs) ===")
print(df_2024[df_2024["m (kg)"] > 0]["m (kg)"].sort_values().unique()[:10])

print("\n=== Sample of Highest 10 Unique Weights (Potential Bugs) ===")
print(df_2024["m (kg)"].sort_values(ascending=False).unique()[:10])
```

## Or, since we know each line in our data represents one single car at a time, and the same car with the same specs is represented in our data multiple times, we could group our data by the RLFI column, and use the mean of the values for the numeric fields:

```python
# 1. We define the aggregation rules to be applied to each column
agg_rules = {
    'Ewltp (g/km)': 'mean',          # Continuous Numeric: Average CO2 emissions
    'Mt': 'mean',                    # Continuous Numeric: Average test weight
    'm (kg)': 'mean',                # Continuous Numeric: Average curb weight
    'ec (cm3)': 'mean',              # Continuous Numeric: Average engine displacement
    'Electric range (km)': 'mean',   # Continuous Numeric: Average electric range
    'r': 'sum',                      # Registration Count: Total number of vehicles sold from that family
    
    # Categorical Columns: Usually the same within the group, we take the first value to be safe
    'Mh': 'first',                   # Standard Manufacturer Name
    'Mp': 'first',                   # Manufacturer Pool
    'Fm': 'first',                   # Fuel Mode (E, P, H, M etc.)
    'Ct': 'first'                    # Vehicle Category (M1, M1G etc.)
}

# 2. We are performing the grouping process
df_grouped = df_de_fr_2024.groupby('RLFI').agg(agg_rules).reset_index()

# 3. We are checking the size of the new dataset
print(f"Old Row Count: {len(df_de_fr_2024)}")
print(f"New Grouped Row Count: {len(df_grouped)}")
```
